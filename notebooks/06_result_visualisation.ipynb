```python
import matplotlib.pyplot as plt

# Data preparation
versions = ['v7', 'v8', 'v9', 'v10']

# Metric scores across versions
grammar_score = [54.5, 61.2, 76.6, 80.2]
meaning_score = [58.4, 71.3, 79.9, 87.3]
strict_accuracy = [46.8, 51.2, 68.3, 77.1]
wer = [13.05, 7.64, 5.26, 3.68]
gleu = [76.98, 83.44, 89.25, 92.14]

# Initialize the plot (made slightly larger to fit text labels better)
plt.figure(figsize=(12, 7))

# Plot each metric with an explicitly assigned color
plt.plot(versions, grammar_score, marker='o', linewidth=2, color='#1f77b4', label='Grammar Score (%)')
plt.plot(versions, meaning_score, marker='s', linewidth=2, color='#ff7f0e', label='Meaning Score (%)')
plt.plot(versions, strict_accuracy, marker='^', linewidth=2, color='#2ca02c', label='STRICT ACCURACY (%)')
plt.plot(versions, gleu, marker='D', linewidth=2, color='#d62728', label='GLEU (%)')
plt.plot(versions, wer, marker='X', linewidth=2, color='#9467bd', label='WER (%)')

# A helper function to add the exact numbers on top of the points
def add_labels(x, y, labels=None):
    for i in range(len(x)):
        # If specific labels are provided (like for GLEU), use those; otherwise use the Y value
        text_val = labels[i] if labels else y[i]
        
        # Add a slight vertical offset (+1.5) so the text sits right above the point
        plt.text(x[i], y[i] + 1.5, str(text_val), ha='center', va='bottom', fontsize=10, fontweight='bold')

# Apply labels to all points
add_labels(versions, grammar_score)
add_labels(versions, meaning_score)
add_labels(versions, strict_accuracy)
add_labels(versions, wer)
add_labels(versions, gleu) 

# Add titles and labels
plt.title('Text-Tune-Small (3b) Performance Metrics: v7 to v10', fontsize=15, pad=15, fontweight='bold')
plt.xlabel('Model Version', fontsize=12)
plt.ylabel('Score (%)', fontsize=12)

# Set the Y-axis limit a bit higher so the top text labels aren't cut off
plt.ylim(0, 105)

# Add grid lines
plt.grid(True, linestyle='--', alpha=0.7)

# Add legend outside the plot
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

# Adjust layout to fit everything
plt.tight_layout()

# Display the plot
plt.show()
```
